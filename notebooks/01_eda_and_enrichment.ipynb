## 1. Schema Understanding & Validation

The unified schema design avoids forcing interpretation onto raw events. Instead:
- `event` records carry neutral categories (`product_launch`, `market_entry`, `policy`, `infrastructure`, etc.) with empty `pillar` fields.
- `impact_link` records connect an event (`parent_id`) to a specific indicator (`related_indicator`) and assign the pillar of the affected indicator, alongside impact estimates and lag durations.
- `observation` records contain measured empirical indicators across standard financial inclusion pillars (`ACCESS`, `USAGE`, `AFFORDABILITY`, `GENDER`, etc.).

```python
impact_links = repo.get_impact_links()
events = repo.get_events()
observations = repo.get_observations()

print(f"Total Observations: {len(observations)}")
print(f"Total Neutral Events: {len(events)}")
print(f"Total Impact Links: {len(impact_links)}")

# Inspect sample impact link
if impact_links:
    sample_imp = impact_links[0]
    print("\nSample Impact Link:")
    for k, v in sample_imp.items():
        if v:
            print(f"  {k}: {v}")
```

## 2. Access & Usage Trajectory (2011–2024)

### Key Insights:
1. **Account Ownership Trajectory (2011–2024)**: Account ownership grew from 14.0% in 2011 to 22.0% in 2014, 35.0% in 2017, 46.0% in 2021, reaching 49.0% in 2024.
2. **The 2021–2024 Growth Slowdown**: Between 2017 and 2021, account ownership surged by +11.0 percentage points (+2.75 pp/year), driven by aggressive traditional bank branch expansion (CBE, Awash, Dashen). However, between 2021 and 2024, growth slowed dramatically to +3.0 percentage points (+1.0 pp/year). This slowdown highlights bank branch saturation in urban centers and high e-KYC friction in rural areas prior to Fayda ID rollout.
3. **Mobile Money Hyper-Growth vs Traditional Banking**: While traditional bank account growth slowed down, Mobile Money Account Rate doubled from 4.7% in 2021 to 9.45% in 2024 following Telebirr's launch (over 54M users) and M-Pesa's entry (over 10.8M users).
4. **Digital Payment Crossover Point**: Instant P2P transactions reached 128.3M transactions (577.7B ETB value) in FY24/25, surpassing physical ATM transactions (119.3M count, 156.1B ETB value) for a P2P/ATM crossover ratio of 1.08.
5. **Gender Inclusion Gap Persistence**: The gender gap in formal account ownership remains wide at 18.0 percentage points in 2024 (down slightly from 20.0 pp in 2021), while female mobile money share stands at 14.0%, underscoring the urgent need for targeted female agent networks and digital literacy.
