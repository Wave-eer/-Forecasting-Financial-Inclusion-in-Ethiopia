## 2. Methodology & Functional Forms

### Adoption & Impact Decay Curves:
1. **Sigmoidal Curve**: $I(t) = \frac{M}{1 + e^{-k(t/L - 0.5)}}$ captures initial setup friction, exponential adoption, and market saturation.
2. **Logarithmic Model**: $I(t) = M \cdot \frac{\ln(1 + 9t/L)}{\ln(10)}$ captures rapid initial user onboarding (e.g., Telebirr 0 to 20M users) followed by gradual growth.
3. **Linear Model**: $I(t) = M \cdot \min(1, t/L)$ used for linear regulatory rollouts.
